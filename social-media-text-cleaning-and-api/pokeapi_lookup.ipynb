```python
# request PokeAPI
# import requests module
import requests

# Making a get request
response = requests.get('https://pokeapi.co/api/v2/')

# Store JSON data in API_Data
API_Data = response.json()

# Print JSON data using loop
for key in API_Data:
    print(key, ":", API_Data[key])
```

```python
def get_pokemon_data(name):
    url = f"https://pokeapi.co/api/v2/pokemon/{name.lower()}"
    response = requests.get(url)
    
    if response.status_code == 200:
        return response.json()
    else:
        print("Pokémon not found.")
        return None


def display_pokemon_info(data):
    # Basic Information
    print(f"\nName: {data['name'].capitalize()}")
    print(f"ID: {data['id']}")
    print(f"Height: {data['height']}")
    print(f"Weight: {data['weight']}")
    
    # Types
    print("\nTypes:")
    for type_info in data['types']:
        print(f"- {type_info['type']['name']}")
    
    # Abilities
    print("\nAbilities:")
    for ability in data['abilities']:
        name = ability['ability']['name']
        is_hidden = ability['is_hidden']
        if is_hidden:
            print(f"- {name} (hidden)")
        else:
            print(f"- {name}")
    
    # Base Stats
    print("\nBase Stats:")
    for stat in data['stats']:
        stat_name = stat['stat']['name'].replace('-', ' ').capitalize()
        stat_value = stat['base_stat']
        print(f"- {stat_name}: {stat_value}")
```

```python
def main():
    while True:
        pokemon_name = input("Enter Pokémon name: ").strip()
        data = get_pokemon_data(pokemon_name)
        
        if data:
            display_pokemon_info(data)
            break
        else:
            print("Please try again. \n")

if __name__ == "__main__":
    main()
```
